```python
import pandas as pd

df1 = pd.DataFrame({
    'Employee': ['Alice', 'Bob'],
    'Q1':[100,120],
    'Q2':[120,140],
    'Q3': [130, 160]
})
```

```python
df1_melted=pd.melt(df1,id_vars=['Employee'],value_vars=['Q1','Q2','Q3'],var_name='Quarter',value_name='Revenue')
```

```python
print(df1_melted)
```

```python
df2 = pd.read_csv('../data/raw/raw_data.csv')
```

```python
df2_melted=pd.melt(df2,id_vars=['Class','Student_ID'],value_name='Score',var_name='Subject')
```

When the value_vars is ignored in the input of pd.melt, the rest of the dataframe will all be melted(relative to the columns we fixed with id_vars)

```python
print(df2_melted)
```

```python
df2_melted.to_csv('../results/processed/cleaned_data.csv', index=False)
```
